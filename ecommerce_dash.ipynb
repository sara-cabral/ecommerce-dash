## 1. Carregamento dos dados

Leitura do arquivo `ecommerce_estatistica.csv` utilizando a biblioteca Pandas.

```python
import pandas as pd
import plotly.express as px
from dash import Dash, dcc, html
```

```python
df = pd.read_csv('ecommerce_estatistica.csv')
df.head()
```

## 2. Criação da aplicação Dash

Nesta etapa é criada a estrutura básica da aplicação que será utilizada para apresentar as visualizações.

```python
app = Dash(__name__)

app.layout = html.Div([
    html.H1('Análise de Dados - E-commerce'),

    dcc.Graph(figure=fig_histograma),

    dcc.Graph(figure=fig_dispersao),

    dcc.Graph(figure=fig_heatmap),

    dcc.Graph(figure=fig_barras),

    dcc.Graph(figure=fig_pizza),

    dcc.Graph(figure=fig_densidade),

    dcc.Graph(figure=fig_regressao)
])
```

## 3. Distribuição dos preços

O histograma permite visualizar como os preços dos produtos estão distribuídos.

```python
fig_histograma = px.histogram(
    df,
    x='Preço_MinMax',
    title='Distribuição dos Preços dos Produtos'
)
```

## 4. Relação entre preço e avaliações

O gráfico de dispersão permite observar a relação entre o preço dos produtos e o número de avaliações.

```python
fig_dispersao = px.scatter(
    df,
    x='Preço_MinMax',
    y='N_Avaliações',
    title='Relação entre Preço e Número de Avaliações'
)
```

## 5. Correlação entre indicadores

O mapa de calor apresenta as correlações entre os principais indicadores numéricos dos produtos.

```python
correlacao = df[['Nota', 'N_Avaliações', 'Desconto',
                 'Preço_MinMax', 'Qtd_Vendidos_Cod']].corr()
```

```python
fig_heatmap = px.imshow(
    correlacao,
    text_auto=True,
    title='Correlação entre Indicadores dos Produtos'
)
```

## 6. Marcas com maior quantidade de produtos

O gráfico de barras apresenta as 10 marcas com maior número de produtos no conjunto de dados.

```python
marcas = df['Marca'].value_counts().head(10)
```

## 6. Marcas com maior quantidade de produtos

O gráfico de barras apresenta as 10 marcas com maior número de produtos no conjunto de dados.

```python
fig_barras = px.bar(
    marcas,
    title='10 Marcas com Maior Número de Produtos',
    labels={
        'index': 'Marca',
        'value': 'Quantidade de Produtos'
    }
)
```

## 7. Distribuição por temporada

O gráfico de pizza apresenta a distribuição dos produtos entre as principais categorias de temporada.

```python
contagem = df['Temporada'].value_counts()

principais = contagem.head(4)
outros = contagem.iloc[4:].sum()

dados_pizza = pd.concat([
    principais,
    pd.Series({'Outros': outros})
])
```

```python
fig_pizza = px.pie(
    dados_pizza,
    values=dados_pizza.values,
    names=dados_pizza.index,
    title='Distribuição dos Produtos por Temporada'
)
```

## 8. Distribuição dos preços

Esta visualização permite observar a distribuição e a concentração dos preços dos produtos.

```python
fig_densidade = px.histogram(
    df,
    x='Preço_MinMax',
    marginal='violin',
    title='Densidade da Distribuição dos Preços'
)
```

## 9. Modelo de regressão

A visualização apresenta a relação entre número de avaliações, preço e quantidade vendida, utilizando uma superfície de regressão para representar a tendência dos dados.

```python
fig_regressao = px.scatter(
    df,
    x='N_Avaliações',
    y='Qtd_Vendidos_Cod',
    trendline='ols',
    title='Relação entre Número de Avaliações e Quantidade Vendida'
)
```

## 10. Execução da aplicação

A aplicação Dash reúne as visualizações desenvolvidas e permite que o usuário explore os gráficos de forma interativa.

```python
app.run(debug=True)
```
